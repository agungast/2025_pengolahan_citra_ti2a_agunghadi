# Dasar Berkas Gambar

```python
# PRAKTIKUM_01_
import numpy as np
import matplotlib.pyplot as plt
from skimage import data

# Mengambil contoh gambar dari scikit-image
gambar_list = [data.coins(), data.camera(), data.astronaut()]
nama_list = ["Coins", "Camera", "Astronaut"]

# Fungsi untuk menghitung ukuran gambar dalam bit, byte, KB, MB
def hitung_ukuran(image):
    bit_per_pixel = 8 if image.ndim == 2 else 24  # Grayscale = 8 bit, RGB = 24 bit
    total_pixels = image.shape[0] * image.shape[1]
    total_bits = total_pixels * bit_per_pixel
    total_bytes = total_bits / 8
    total_kb = total_bytes / 1024
    total_mb = total_kb / 1024
    return total_bits, total_bytes, total_kb, total_mb

# Menampilkan gambar dan ukurannya
fig, axes = plt.subplots(1, 3, figsize=(15, 5))

for ax, img, nama in zip(axes, gambar_list, nama_list):
    bits, bytes_, kb, mb = hitung_ukuran(img)
    ax.imshow(img, cmap='gray' if img.ndim == 2 else None)
    ax.set_title(f"{nama}\n{bits} bit, {bytes_:.2f} B\n{kb:.2f} KB, {mb:.4f} MB")
    ax.axis('off')

plt.tight_layout()
plt.show()
```

```python
# PRAKTIKUM_02_Menampilkan bagaimana citra digital tersusun dari piksel dengan area fokus yang ditandai
import numpy as np
import matplotlib.pyplot as plt
from skimage import data

# Mengambil contoh gambar dari scikit-image
gambar_list = [data.coins(), data.camera(), data.astronaut()]
nama_list = ["Coins", "Camera", "Astronaut"]

# Koordinat area kecil yang di-zoom
x_start, x_end = 50, 60
y_start, y_end = 50, 60

# Menampilkan bagaimana citra digital tersusun dari piksel dengan area fokus yang ditandai
fig, axes = plt.subplots(3, 3, figsize=(12, 12))

for i, (img, nama) in enumerate(zip(gambar_list, nama_list)):
    # Gambar asli
    axes[i, 0].imshow(img, cmap='gray' if img.ndim == 2 else None)
    axes[i, 0].set_title(f"{nama} - Gambar Asli")
    axes[i, 0].axis('off')

    # Tambahkan kotak merah untuk menunjukkan area yang difokuskan
    rect_x = [x_start, x_end, x_end, x_start, x_start]
    rect_y = [y_start, y_start, y_end, y_end, y_start]
    axes[i, 0].plot(rect_x, rect_y, color="red", linewidth=2)

    # Gambar fokus area
    zoomed_img = img[y_start:y_end, x_start:x_end]
    axes[i, 1].imshow(zoomed_img, cmap='gray' if img.ndim == 2 else None, interpolation='nearest')
    axes[i, 1].set_title("Gambar Fokus Area")
    axes[i, 1].axis('off')

    # Menampilkan nilai piksel dalam area yang di-zoom
    axes[i, 2].imshow(zoomed_img, cmap='gray' if img.ndim == 2 else None, interpolation='nearest')

    for y in range(zoomed_img.shape[0]):
        for x in range(zoomed_img.shape[1]):
            pixel_value = np.mean(zoomed_img[y, x]).astype(int) if img.ndim == 3 else zoomed_img[y, x]
            text_color = "white" if pixel_value < 128 else "black"
            axes[i, 2].text(x, y, str(pixel_value), color=text_color, ha='center', va='center', fontsize=10, fontweight='bold')

    axes[i, 2].set_title("Nilai Piksel (Mudah Dibaca)")
    axes[i, 2].axis('off')

plt.tight_layout()
plt.show()
```

```python
# PRAKTIKUM_03_Menjelaskan bagaimana citra disimpan dalam array NumPy
import numpy as np
import matplotlib.pyplot as plt
from skimage import data

# Mengambil contoh gambar dari scikit-image
gambar_list = [data.coins(), data.camera(), data.astronaut()]
nama_list = ["Coins", "Camera", "Astronaut"]

# Fungsi untuk menampilkan informasi array NumPy dari gambar
def info_array_numpy(image, nama):
    print(f"\n=== {nama} ===")
    print(f"Tipe Data: {type(image)}")  # Harusnya <class 'numpy.ndarray'>
    print(f"Dimensi: {image.shape}")  # Menampilkan dimensi gambar
    print(f"Tipe Nilai Piksel: {image.dtype}")  # Menampilkan tipe data dalam array
    print("Contoh nilai piksel (5x5 pertama):")
    print(image[:5, :5] if image.ndim == 2 else image[:5, :5, :])

# Menampilkan informasi setiap gambar
for img, nama in zip(gambar_list, nama_list):
    info_array_numpy(img, nama)

# Visualisasi bagaimana array NumPy mewakili citra
fig, axes = plt.subplots(1, 3, figsize=(15, 5))

for ax, img, nama in zip(axes, gambar_list, nama_list):
    ax.imshow(img, cmap='gray' if img.ndim == 2 else None)
    ax.set_title(f"{nama}\nArray Shape: {img.shape}")
    ax.axis('off')

plt.tight_layout()
plt.show()
```

```python
# PRAKTIKUM_04_ Menjelaskan sistem koordinat kiri atas dalam citra digital dengan warna anotasi yang lebih kontras
import numpy as np
import matplotlib.pyplot as plt
from skimage import data

# Pilih salah satu gambar untuk contoh koordinat
gambar_sample = data.coins()

# Ukuran gambar
height, width = gambar_sample.shape

# Buat plot
fig, ax = plt.subplots(figsize=(6, 6))

# Tampilkan gambar
ax.imshow(gambar_sample, cmap='gray')

# Tambahkan anotasi sumbu koordinat
ax.set_title("Sistem Koordinat Kiri Atas dalam Citra Digital", fontsize=12, fontweight="bold")
ax.set_xlabel("Sumbu X (Lebar) →", fontsize=10, color="blue")
ax.set_ylabel("↓ Sumbu Y (Tinggi)", fontsize=10, color="red")

# Tambahkan garis koordinat utama
ax.axhline(y=0, color='red', linestyle='--', linewidth=1, label="Y = 0 (Atas)")
ax.axvline(x=0, color='blue', linestyle='--', linewidth=1, label="X = 0 (Kiri)")

# Tambahkan beberapa titik koordinat penting dengan warna kontras (cyan)
koordinat_titik = [
    (0, 0), (width - 1, 0), (0, height - 1), (width - 1, height - 1), (width // 2, height // 2)
]

for x, y in koordinat_titik:
    ax.scatter(x, y, color='cyan', s=50, edgecolors='black', linewidth=1.2)  # Titik dengan outline hitam
    ax.text(x + 10, y + 10, f"({x}, {y})", color="cyan", fontsize=10, fontweight="bold", ha="left", va="bottom")

# Tambahkan legenda
ax.legend(loc="upper right")

plt.show()
```

```python
# PRAKTIKUM_05_ Menjelaskan model warna aditif RGB dalam citra digital dengan RGB Triplet dan warna hitam
import numpy as np
import matplotlib.pyplot as plt

# Membuat tiga kanal warna utama dalam model RGB
red_channel = np.zeros((100, 100, 3), dtype=np.uint8)
green_channel = np.zeros((100, 100, 3), dtype=np.uint8)
blue_channel = np.zeros((100, 100, 3), dtype=np.uint8)

# Mengatur warna merah, hijau, dan biru murni
red_channel[:, :, 0] = 255    # Merah (255, 0, 0)
green_channel[:, :, 1] = 255  # Hijau (0, 255, 0)
blue_channel[:, :, 2] = 255   # Biru (0, 0, 255)

# Membuat kombinasi warna dari RGB
yellow = red_channel + green_channel    # Kuning (255, 255, 0)
cyan = green_channel + blue_channel     # Cyan (0, 255, 255)
magenta = red_channel + blue_channel    # Magenta (255, 0, 255)
white = red_channel + green_channel + blue_channel  # Putih (255, 255, 255)
black = np.zeros((100, 100, 3), dtype=np.uint8)  # Hitam (0, 0, 0)

# Menampilkan hasil dengan RGB Triplet
fig, axes = plt.subplots(2, 4, figsize=(12, 6))

# List warna, nama, dan RGB triplet
color_list = [red_channel, green_channel, blue_channel, white, yellow, cyan, magenta, black]
color_names = ["Merah", "Hijau", "Biru", "Putih", "Kuning", "Cyan", "Magenta", "Hitam"]
rgb_values = ["(255, 0, 0)", "(0, 255, 0)", "(0, 0, 255)", "(255, 255, 255)",
              "(255, 255, 0)", "(0, 255, 255)", "(255, 0, 255)", "(0, 0, 0)"]

# Plot warna dengan keterangan RGB triplet
for ax, color, name, rgb in zip(axes.flat, color_list, color_names, rgb_values):
    ax.imshow(color)
    ax.set_title(f"{name}\nRGB {rgb}", fontsize=10, fontweight="bold")
    ax.axis('off')

plt.suptitle("Model Warna Aditif RGB dengan RGB Triplet & Hitam", fontsize=14, fontweight="bold")
plt.tight_layout()
plt.show()
```

```python
# PRAKTIKUM_06_Menjelaskan urutan tiga nilai warna dalam citra scikit-image dengan titik penanda koordinat piksel
import numpy as np
import matplotlib.pyplot as plt
from skimage import data

# Memuat gambar RGB dari scikit-image
gambar_rgb = data.astronaut()

# Ekstrak tiga kanal warna (R, G, B)
red_channel = gambar_rgb[:, :, 0]    # Kanal merah
green_channel = gambar_rgb[:, :, 1]  # Kanal hijau
blue_channel = gambar_rgb[:, :, 2]   # Kanal biru

# Koordinat tiga titik sampel untuk menampilkan nilai piksel RGB
titik_koordinat = [(100, 100), (200, 150), (300, 250)]

# Menampilkan gambar asli dan masing-masing kanal warna
fig, axes = plt.subplots(1, 4, figsize=(16, 4))

# Daftar kanal warna dan colormap yang sesuai
kanal_list = [gambar_rgb, red_channel, green_channel, blue_channel]
judul_list = ["Gambar Asli (RGB)", "Kanal Merah (Red)", "Kanal Hijau (Green)", "Kanal Biru (Blue)"]
colormap_list = [None, "Reds", "Greens", "Blues"]

# Plot setiap gambar dengan titik penanda
for ax, img, title, cmap in zip(axes, kanal_list, judul_list, colormap_list):
    ax.imshow(img, cmap=cmap)
    ax.set_title(title, fontsize=10, fontweight="bold")
    ax.axis("off")

    # Tambahkan titik sampel
    for x, y in titik_koordinat:
        ax.scatter(x, y, color='yellow' if cmap is None else 'black', s=50, edgecolors='white', linewidth=1.2)

plt.suptitle("Urutan Tiga Nilai Warna dalam Citra dengan Titik Sampel", fontsize=14, fontweight="bold")
plt.tight_layout()
plt.show()

# Menampilkan nilai RGB pada titik-titik sampel dengan format yang diinginkan
print("Nilai RGB pada titik koordinat yang dipilih:")
for i, (x, y) in enumerate(titik_koordinat, start=1):
    r_val, g_val, b_val = gambar_rgb[y, x]
    print(f"Titik {i} - Koordinat ({x}, {y}):")
    print(f"  Red   (Merah)   : {r_val}")
    print(f"  Green (Hijau)   : {g_val}")
    print(f"  Blue  (Biru)    : {b_val}")
    print(f"  RGB Triplet     : ({r_val}, {g_val}, {b_val})\n")
```

```python
# PRAKTIKUM_07_Karakteristik format citra BMP, JPEG, dan TIFF
import os
import numpy as np
import matplotlib.pyplot as plt
from skimage import io, data

# Memuat gambar contoh dari skimage
gambar = data.astronaut()

# Simpan gambar dalam format BMP, JPEG, dan TIFF
io.imsave("astronaut.bmp", gambar)
io.imsave("astronaut.jpg", gambar, quality=95)  # JPEG dengan kualitas tinggi
io.imsave("astronaut.tiff", gambar)

# Fungsi untuk menghitung ukuran file
def ukuran_file(nama_file):
    size_bytes = os.path.getsize(nama_file)
    size_kb = size_bytes / 1024
    size_mb = size_kb / 1024
    return f"{size_bytes} Bytes ({size_kb:.2f} KB, {size_mb:.4f} MB)"

# Daftar format dan nama file
format_citra = ["BMP", "JPEG", "TIFF"]
nama_file = ["astronaut.bmp", "astronaut.jpg", "astronaut.tiff"]
ukuran = [ukuran_file(file) for file in nama_file]

# Tampilkan gambar dan informasi karakteristik format citra
fig, axes = plt.subplots(1, 3, figsize=(15, 5))

for ax, fmt, file in zip(axes, format_citra, nama_file):
    img = io.imread(file)
    ax.imshow(img)
    ax.set_title(f"{fmt}\nUkuran:\n{ukuran_file(file)}", fontsize=12, fontweight="bold")
    ax.axis("off")

plt.suptitle("Karakteristik Format Citra BMP, JPEG, dan TIFF", fontsize=16, fontweight="bold", color="darkblue", y=1.05)
plt.tight_layout()
plt.show()

# Menampilkan informasi karakteristik format citra
print("Karakteristik Format Citra:")

info_format = [
    ("BMP (Bitmap):",
     "- Format tidak terkompresi, menghasilkan ukuran file yang besar.\n"
     "- Kualitas gambar sangat baik, tidak ada kehilangan data.\n"
     "- Cocok untuk pengolahan citra yang memerlukan ketelitian tinggi.",
     ukuran[0]),

    ("JPEG (Joint Photographic Experts Group):",
     "- Format terkompresi dengan metode lossy (menghilangkan sebagian data).\n"
     "- Ukuran file lebih kecil dibanding BMP dan TIFF.\n"
     "- Cocok untuk fotografi, media sosial, dan tampilan web.",
     ukuran[1]),

    ("TIFF (Tagged Image File Format):",
     "- Format fleksibel yang mendukung kompresi lossy atau lossless.\n"
     "- Digunakan dalam pencetakan profesional dan pemrosesan citra medis.\n"
     "- Ukuran file bisa besar tergantung tingkat kompresi yang digunakan.",
     ukuran[2])
]

for nama, deskripsi, ukuran_file in info_format:
    print(f"\n{nama}")
    print(deskripsi)
    print(f"- Ukuran file: {ukuran_file}")
```

```python
# PRAKTIKUM_08_ Menjelaskan perbedaan kompresi lossy dan lossless dalam citra digital
import os
import matplotlib.pyplot as plt
from skimage import io, data

# Memuat gambar contoh dari skimage
gambar = data.astronaut()

# Simpan gambar dalam format lossy (JPEG) dan lossless (PNG)
jpeg_quality_50 = "astronaut_lossy_50.jpg"
jpeg_quality_90 = "astronaut_lossy_90.jpg"
png_lossless = "astronaut_lossless.png"

io.imsave(jpeg_quality_50, gambar, quality=50)  # JPEG lossy dengan kualitas rendah
io.imsave(jpeg_quality_90, gambar, quality=90)  # JPEG lossy dengan kualitas tinggi
io.imsave(png_lossless, gambar)  # PNG lossless

# Fungsi untuk menghitung ukuran file
def ukuran_file(nama_file):
    size_bytes = os.path.getsize(nama_file)
    size_kb = size_bytes / 1024
    size_mb = size_kb / 1024
    return f"{size_bytes} Bytes ({size_kb:.2f} KB, {size_mb:.4f} MB)"

# Menampilkan ukuran file dari masing-masing format
nama_file_kompresi = [jpeg_quality_50, jpeg_quality_90, png_lossless]
ukuran_kompresi = [ukuran_file(file) for file in nama_file_kompresi]

# Menampilkan gambar hasil kompresi
fig, axes = plt.subplots(1, 3, figsize=(15, 5))
kompresi_label = ["JPEG (Lossy, Q=50)", "JPEG (Lossy, Q=90)", "PNG (Lossless)"]

for ax, file, label in zip(axes, nama_file_kompresi, kompresi_label):
    img = io.imread(file)
    ax.imshow(img)
    ax.set_title(f"{label}\nUkuran:\n{ukuran_file(file)}", fontsize=12, fontweight="bold")
    ax.axis("off")

plt.suptitle("Perbedaan Kompresi Lossy dan Lossless dalam Citra Digital", fontsize=16, fontweight="bold", color="darkblue")
plt.tight_layout()
plt.show()

# Menampilkan informasi perbedaan kompresi dalam teks
print("Perbedaan Kompresi Lossy dan Lossless:")

info_kompresi = [
    ("JPEG (Lossy, Quality=50):",
     "- Menggunakan kompresi lossy dengan kualitas rendah (Q=50).\n"
     "- Detail gambar berkurang, muncul artefak kompresi.\n"
     "- Ukuran file lebih kecil.",
     ukuran_kompresi[0]),

    ("JPEG (Lossy, Quality=90):",
     "- Menggunakan kompresi lossy dengan kualitas lebih tinggi (Q=90).\n"
     "- Detail gambar masih cukup baik, artefak lebih sedikit.\n"
     "- Ukuran file lebih besar dibanding Q=50, tetapi lebih kecil dibanding lossless.",
     ukuran_kompresi[1]),

    ("PNG (Lossless):",
     "- Menggunakan kompresi lossless, tidak ada kehilangan data.\n"
     "- Detail gambar tetap sempurna seperti aslinya.\n"
     "- Ukuran file lebih besar dibanding JPEG lossy.",
     ukuran_kompresi[2])
]

for nama, deskripsi, ukuran_file in info_kompresi:
    print(f"\n{nama}")
    print(deskripsi)
    print(f"- Ukuran file: {ukuran_file}")
```
